## Function 1 - torch.resize_

`resize_(*sizes, memory_format=torch.contiguous_format) → Tensor`

This function resizes a tensor to the specified size. If the specified size is too small, the new tensor will remove some elements. If the specified size is too large, the tensor will not initialize the remaining empty element slots

<h3> Example 1 - specified *size can accomodate all elements </h3>

```python
a = torch.tensor([[1., 2., 7.], [3., 4., 0.], [5., 6., 14.], [7., 8., 10.]])
print("Original tensor: \n", a)
print("\nOriginal tensor shape: ", a.shape)

b = a.resize_(6,2)
print("\nResized tensor: \n", b)
print("\nResized tensor shape: ", b.shape)
```

In the above example, a tensor, `a`, has 12 elements and is of shape 4x3.
<br>
The new tensor, `b`, is obtained by resizing `a` into a 6x2 tensor = 12 elements

<h3> Example 2 - specified *size exceeds the number of elements </h3>

```python
a = torch.tensor([[1.], [3.], [5.], [7.]])
print("Original tensor: \n", a)
print("\nOriginal tensor shape: ", a.shape)

b = a.resize_(2,4)
print("\nResized tensor: \n", b)
print("\nResized tensor shape: ", b.shape)
```

In the above example, a tensor, `a`, has 4 elements and is of shape 4x1.
<br>
The new tensor, `b`, is obtained by resizing `a` into a 2x4 tensor. As `b` can contain 8 elements, the remaining 4 elements are initialized to 0.

<h3> Example 3 - specified *size is lower than the number of elements </h3>

```python
a = torch.tensor([[1., 2., 7.], [3., 4., 0.], [5., 6., 14.], [7., 8., 10.]])
print("Original tensor: \n", a)
print("\nOriginal tensor shape: ", a.shape)
print("\nNumber of elements: ", a.numel())

b = a.resize_(5,2)
print("\nResized tensor: \n", b)
print("\nResized tensor shape: ", b.shape)
print("\nNumber of elements: ", b.numel())
```

In the above example, a tensor, `a`, has 12 elements and is of shape 4x3.
<br>
The new tensor, `b`, is obtained by resizing `a` into a 5x2 tensor. As `b` can contain 10 elements, the remaining last 2 elements of `a` are dropped.

`.resize_` should be used when there is a need to resize a tensor to a new specified dimension. <br>
This function doesn't usually break

## Function 2 - torch.argmin

`torch.argmin(input) → LongTensor`

This function returns the indices of the minimum value of all elements in the `input` tensor.

<br>

<h3> Example 1 - Returns the index of the minimum element in the whole tensor </h3>

```python
# Generates a random tensor of size 5x6
a = torch.randn(5, 6)
print("Created tensor: \n", a)
print("Index of smallest element is: ", torch.argmin(a))
```

```python
# Access the smallest element by index by resizing the index into a vector
b = a.resize_(1, a.numel())
print("Smallest element is: ", b[0][torch.argmin(a)])
```

In the above sample, `torch.argmin(input_tensor)` is used to obtain the smallest element in the whole tensor, from the starting 0th element. From this, the tensor is resized to a vector and the smallest element is extracted. Alternatively, `torch.min(input_tensor)` can also be used

```python
print("Smallest element is: ", torch.min(a))
```

<h3> Example 2 (Breaking) - Only returns a single index when all the elements are the same value </h3>

```python
a = torch.tensor([[1., 5., 1.], [3., 1., 7.]])
torch.argmin(a)
```

When there are multiple elements with the same, minimum value, only a single index is returned. To rectify this, we can do another look and check all values which have the same minimum value.

```python
min_value = a.resize_(1, a.numel())[0][torch.argmin(a)]
print ((a == min_value.item()).nonzero())
```

In the above example, the min_value is first obtained. Following which, `min_value.item()` is used to obtain the number itself from the 1-D tensor, which is used for comparison. <br>
The index of all the elements which has the minimum value is then returned in the subsequent tensor
